```python
import pandas as pd
```

```python
df_insee = pd.read_csv(
    "../Data/Input/base-cc-logement-2022.CSV",
    sep=";",
    dtype={"CODGEO": str}
)
```

```python
df_insee.shape
```

```python
df_insee_95 = df_insee[
    df_insee["CODGEO"].str.startswith("95")
].copy()

print("Dimensions Val-d'Oise :", df_insee_95.shape)
```

```python
colonnes_utiles = [
    "CODGEO",

    # Taille et occupation du parc
    "P22_LOG",
    "P22_RP",
    "P22_RSECOCC",
    "P22_LOGVAC",

    # Type de logements
    "P22_MAISON",
    "P22_APPART",

    # Ancienneté des résidences principales
    "P22_RP_ACH1919",
    "P22_RP_ACH1945",
    "P22_RP_ACH1970",
    "P22_RP_ACH1990",
    "P22_RP_ACH2005",
    "P22_RP_ACH2019"
]

df_insee_95 = df_insee_95[colonnes_utiles].copy()

print("Dimensions :", df_insee_95.shape)
```

```python
print("Types de données :")
print(df_insee_95.dtypes)

print("\nValeurs manquantes :")
print(df_insee_95.isna().sum())
```

```python
print("Nombre de lignes :", len(df_insee_95))
print("Codes INSEE uniques :", df_insee_95["CODGEO"].nunique())
print("Doublons sur CODGEO :", df_insee_95["CODGEO"].duplicated().sum())
```

```python
df_insee_95 = df_insee_95.rename(columns={
    "CODGEO": "code_insee",

    "P22_LOG": "logements_total_2022",
    "P22_RP": "residences_principales_2022",
    "P22_RSECOCC": "residences_secondaires_occasionnelles_2022",
    "P22_LOGVAC": "logements_vacants_2022",

    "P22_MAISON": "maisons_2022",
    "P22_APPART": "appartements_2022",

    "P22_RP_ACH1919": "rp_avant_1919",
    "P22_RP_ACH1945": "rp_1919_1945",
    "P22_RP_ACH1970": "rp_1946_1970",
    "P22_RP_ACH1990": "rp_1971_1990",
    "P22_RP_ACH2005": "rp_1991_2005",
    "P22_RP_ACH2019": "rp_2006_2019"
})

df_insee_95.head()
```

```python
df_insee_95.to_csv(
    "../Data/Output/insee_logement_val_doise_2022.csv",
    index=False
)

print("Export terminé : insee_logement_val_doise_2022.csv")
```
