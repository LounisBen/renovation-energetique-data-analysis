```python
import requests
import pandas as pd
```

```python
departement = "95"
```

```python
url_api = (
    "https://data.ademe.fr/data-fair/api/v1/"
    "datasets/liste-des-entreprises-rge-2/lines"
)

params = {
    "code_postal_starts": departement,
    "size": 5
}
```

```python
response = requests.get(
    url_api,
    params=params
)

if response.status_code == 200:
    print("Connexion réussie - code 200")
else:
    print("Erreur :", response.status_code)
```

```python
data = response.json()

print(data.keys())
```

```python
print("Nombre total de lignes trouvées :", data["total"])

df_test = pd.DataFrame(data["results"])

df_test
```

```python
params = {
    "code_postal_starts": departement,
    "size": 10000
}

response = requests.get(
    url_api,
    params=params
)

data = response.json()

df_rge_api = pd.DataFrame(
    data["results"]
)

print("Nombre de lignes récupérées :", len(df_rge_api))
print("Nombre total annoncé par l'API :", data["total"])
```

```python
print("Dimensions :", df_rge_api.shape)

print("\nColonnes :")
print(df_rge_api.columns.tolist())
```

```python
nom_fichier = f"../Data/Input/rge_api_{departement}_raw.csv"

df_rge_api.to_csv(
    nom_fichier,
    index=False
)

print("Fichier exporté :", nom_fichier)
```

```python
colonnes_utiles = [
    "siret",
    "nom_entreprise",
    "adresse",
    "code_postal",
    "commune",
    "code_qualification",
    "nom_qualification",
    "domaine",
    "meta_domaine",
    "organisme",
    "lien_date_debut",
    "lien_date_fin"
]

df_rge_api_clean = df_rge_api[colonnes_utiles].copy()

print("Dimensions :", df_rge_api_clean.shape)
df_rge_api_clean.head()
```

```python
# Les identifiants sont convertis en texte
df_rge_api_clean["siret"] = df_rge_api_clean["siret"].astype(str)
df_rge_api_clean["code_postal"] = df_rge_api_clean["code_postal"].astype(str)

# Conversion des dates
df_rge_api_clean["lien_date_debut"] = pd.to_datetime(
    df_rge_api_clean["lien_date_debut"],
    errors="coerce"
)

df_rge_api_clean["lien_date_fin"] = pd.to_datetime(
    df_rge_api_clean["lien_date_fin"],
    errors="coerce"
)

print(df_rge_api_clean.dtypes)
```

```python
nom_fichier_clean = f"../Data/Output/rge_api_{departement}_clean.csv"

df_rge_api_clean.to_csv(
    nom_fichier_clean,
    index=False
)

print("Fichier exporté :", nom_fichier_clean)
```
