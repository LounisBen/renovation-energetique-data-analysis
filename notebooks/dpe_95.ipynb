```python
import pandas as pd
```

```python
df_dpe_2024 = pd.read_csv(
    "../Data/Input/dpe_95_2024.csv"
)

df_dpe_2025 = pd.read_csv(
    "../Data/Input/dpe_95_2025.csv"
)
```

```python
print("2024 :", df_dpe_2024.shape)
print("2025 :", df_dpe_2025.shape)
```

```python
print(df_dpe_2024.columns.tolist())
```

```python
df_dpe_2024["annee_source"] = 2024
df_dpe_2025["annee_source"] = 2025

df_dpe = pd.concat(
    [df_dpe_2024, df_dpe_2025],
    ignore_index=True
)

print(df_dpe.shape)
```

```python
df_dpe = df_dpe.rename(
    columns={
        "qualite_isolation_plancher bas":
        "qualite_isolation_plancher_bas"
    }
)
```

```python
df_dpe.info()
(df_dpe.isna().mean() * 100).sort_values(ascending=False)
```

```python
df_dpe["code_postal_ban"] = df_dpe["code_postal_ban"].astype(str)
df_dpe["code_insee_ban"] = df_dpe["code_insee_ban"].astype(str)
df_dpe["code_departement_ban"] = df_dpe["code_departement_ban"].astype(str)

df_dpe["date_etablissement_dpe"] = pd.to_datetime(
    df_dpe["date_etablissement_dpe"],
    errors="coerce"
)
```

```python
print("Nombre de lignes :", len(df_dpe))
print("Nombre de DPE uniques :", df_dpe["numero_dpe"].nunique())
print("Doublons :", df_dpe["numero_dpe"].duplicated().sum())
```

```python
colonnes_utiles = [
    "numero_dpe",
    "date_etablissement_dpe",
    "annee_source",
    "etiquette_dpe",
    "etiquette_ges",
    "conso_5_usages_par_m2_ep",
    "type_batiment",
    "surface_habitable_logement",
    "periode_construction",
    "nom_commune_ban",
    "code_postal_ban",
    "code_insee_ban",
    "qualite_isolation_enveloppe",
    "qualite_isolation_murs",
    "qualite_isolation_menuiseries",
    "qualite_isolation_plancher_bas",
    "type_energie_principale_chauffage",
    "type_ventilation"
]

df_dpe_clean = df_dpe[colonnes_utiles].copy()

print(df_dpe_clean.shape)
```

```python
df_dpe_clean = df_dpe_clean.rename(
    columns={
        "nom_commune_ban": "commune",
        "code_postal_ban": "code_postal",
        "code_insee_ban": "code_insee"
    }
)

df_dpe_clean.head()
```

```python
# Repérer les codes INSEE associés à plusieurs écritures de commune
communes_uniques = (
    df_dpe_clean[["code_insee", "commune"]]
    .drop_duplicates()
)

communes_variantes = communes_uniques[
    communes_uniques["code_insee"].duplicated(keep=False)
].sort_values(["code_insee", "commune"])

communes_variantes
```

```python
correction_communes = {
    "BouffÃ mont": "Bouffémont",
    "Bray-et-LÃ": "Bray-et-Lû",
    "CondÃ court": "Condécourt",
    "Ã couen": "Écouen",
    "Ã piais-Rhus": "Épiais-Rhus",
    "Ã ragny": "Éragny",
    "Ã zanville": "Ézanville",
    "FrÃ mainville": "Frémainville",
    "Garges-lÃ¨s-Gonesse": "Garges-lès-Gonesse",
    "MÃ ry-sur-Oise": "Méry-sur-Oise",
    "Montigny-lÃ¨s-Cormeilles": "Montigny-lès-Cormeilles",
    "Saint-Leu-la-Foret": "Saint-Leu-la-Forêt",
    "Saint-Leu-la-ForÃ t": "Saint-Leu-la-Forêt",
    "Saint-Ouen-l'AumÃ ne": "Saint-Ouen-l'Aumône",
    "VÃ theuil": "Vétheuil"
}

df_dpe_clean["commune"] = (
    df_dpe_clean["commune"]
    .replace(correction_communes)
)
```

```python
controle_communes = (
    df_dpe_clean
    .groupby("code_insee")["commune"]
    .nunique()
)

print("Codes avec plusieurs noms :", (controle_communes > 1).sum())
print("Nombre de codes INSEE :", df_dpe_clean["code_insee"].nunique())
print("Nombre de lignes DPE :", len(df_dpe_clean))
```

```python
distribution_dpe = (
    df_dpe_clean["etiquette_dpe"]
    .value_counts()
    .sort_index()
)

print(distribution_dpe)

print(
    (
        df_dpe_clean["etiquette_dpe"]
        .value_counts(normalize=True)
        .sort_index() * 100
    ).round(2)
)
```

```python
df_dpe_clean["passoire_fg"] = (
    df_dpe_clean["etiquette_dpe"]
    .isin(["F", "G"])
).astype(int)

df_dpe_clean["passoire_fg"].value_counts()
```

```python
df_dpe_commune = (
    df_dpe_clean
    .groupby(["code_insee", "commune"])
    .agg(
        nombre_dpe=("numero_dpe", "count"),
        nombre_passoires_fg=("passoire_fg", "sum"),
        consommation_moyenne=(
            "conso_5_usages_par_m2_ep",
            "mean"
        )
    )
    .reset_index()
)

df_dpe_commune["part_passoires_fg_pct"] = (
    df_dpe_commune["nombre_passoires_fg"]
    / df_dpe_commune["nombre_dpe"]
    * 100
).round(2)

df_dpe_commune.head()
```

```python
df_insee_95 = pd.read_csv(
    "../Data/Output/insee_logement_val_doise_2022.csv",
    dtype={"code_insee": str}
)
```

```python
print(
    "Communes DPE :",
    df_dpe_commune["code_insee"].nunique()
)

print(
    "Communes INSEE :",
    df_insee_95["code_insee"].nunique()
)
```

```python
codes_dpe = set(df_dpe_commune["code_insee"])
codes_insee = set(df_insee_95["code_insee"])

print(
    "Présents dans DPE mais pas INSEE :",
    codes_dpe - codes_insee
)

print(
    "Présents dans INSEE mais pas DPE :",
    codes_insee - codes_dpe
)
```
